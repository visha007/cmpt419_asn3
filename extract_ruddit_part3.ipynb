```python
import pandas as pd
import numpy as np
import random

```

```python
rand_sample_count = 10 # The number of random samples we want from our dataset
file_loc="ruddit_comments_score.csv" #The file path to the data you want to sample
the_sheet_name="ruddit_rand_samples_part3.xlsx" #The Excel Sheet name we want to have for our randum samples
rand_samples = []
rand_indecies = []

data=pd.read_csv(file_loc,header=0)
data = data[data["body"]!="[deleted]"] #Remove all of the samples that have [deleted] as their body text
data.reset_index()

total_row_count = data.count()
total_row_count = total_row_count["body"]
print(f"Total Rows = {total_row_count}")

max_index=data.index.max()
min_index=data.index.min()
print(f" Index(min,max) = ({min_index, max_index})")
print(data)

for i in range(rand_sample_count):
    sample_index=random.randint(min_index,max_index)
    rand_indecies.append(sample_index)
    print(f"sampel_index={sample_index}")
    id=data["comment_id"][sample_index]
    text=data["body"][sample_index]
    score=data["score"][sample_index]
    rand_sample = {
        "comment_id":id,
        "text":text,
        "score":score
    }
    rand_samples.append(rand_sample)
    print(rand_sample)

print(rand_sample)

samples_df = pd.DataFrame(rand_samples)

samples_df.to_excel(the_sheet_name,header=True, index=False)

```
